```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
path = '/Users/sandrareinert/Library/Mobile Documents/com~apple~CloudDocs/Documents/Postdoc_SWC/Local_bits/mini2p test/stripes/test2_detrend_suite2p/suite2p/plane0'
settings = 'settings.npy'

ops = 'ops.npy'
```

```python
settings_file = np.load(os.path.join(path, settings), allow_pickle=True).item()
ops_file = np.load(os.path.join(path, ops), allow_pickle=True).item()
```

```python
img1 = ops_file['meanImg']
img2 = ops_file['meanImg_chan2']
#make an image overlay with the two channels
img1 = np.squeeze(img1)
img2 = np.squeeze(img2)
# Normalize images
img1 = (img1 - np.min(img1)) / (np.max(img1) - np.min(img1))
img2 = (img2 - np.min(img2)) / (np.max(img2) - np.min(img2))
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))
ax1.imshow(img1, cmap='gray')
ax2.imshow(img2, cmap='gray')   
```

```python
save_ch1_name = 'meanImg_ch1.tiff'
plt.imsave(os.path.join(path, save_ch1_name), img1, cmap='gray')

save_ch2_name = 'meanImg_ch2.tiff'
plt.imsave(os.path.join(path, save_ch2_name), img2, cmap='gray')
```
